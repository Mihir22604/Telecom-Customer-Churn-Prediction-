## 4. Key Data Cleaning Actions & Results

1. **Target Validation**: Dropped 72 rows missing target `Churn`.
2. **Duplicate Removal**: Dropped 20 exact duplicate rows.
3. **Type Coercion**: Converted `TotalCharges` from `object` to `float64`, replacing 11 `' '` whitespace strings with `0.0` (matching `tenure == 0.0`).
4. **Imputation**: Continuous missing values imputed with Median; categorical missing values imputed with Mode.
5. **Cleaned Output**: Saved 7,019 rows x 21 columns dataset with zero NaNs to `data/processed/cleaned_churn_data.csv`.
